```python
import math, random

import gym
import numpy as np

import torch
import torch.nn as nn
import torch.optim as optim
import torch.autograd as autograd
import torch.nn.functional as F
from torch.autograd import Variable
```

```python
from IPython.display import clear_output
import matplotlib.pyplot as plt
%matplotlib inline
```

<h2>Replay Buffer</h2>

```python
from collections import deque

class ReplayBuffer(object):
    def __init__(self, capacity):
        self.buffer = deque(maxlen=capacity)

    def push(self, state, action, reward, next_state, done):
        state      = np.expand_dims(state, 0)
        next_state = np.expand_dims(next_state, 0)

        self.buffer.append((state, action, reward, next_state, done))

    def sample(self, batch_size):
        state, action, reward, next_state, done = zip(*random.sample(self.buffer, batch_size))
        return np.concatenate(state), action, reward, np.concatenate(next_state), done

    def __len__(self):
        return len(self.buffer)
```

<h2>Cart Pole Environment</h2>

```python
env_id = "CartPole-v0"
env = gym.make(env_id)
```

<h2>Epsilon greedy exploration</h2>

```python
epsilon_start = 1.0
epsilon_final = 0.01
epsilon_decay = 500

epsilon_by_frame = lambda frame_idx: epsilon_final + (epsilon_start - epsilon_final) * math.exp(-1. * frame_idx / epsilon_decay)
```

```python
plt.plot([epsilon_by_frame(i) for i in range(10000)])
```

<h2>Deep Q Network</h2>

```python
class DQN(nn.Module):
    def __init__(self, num_inputs, num_actions):
        super(DQN, self).__init__()

        self.layers = nn.Sequential(
            nn.Linear(env.observation_space.shape[0], 128),
            nn.ReLU(),
            nn.Linear(128, 128),
            nn.ReLU(),
            nn.Linear(128, env.action_space.n)
        )

    def forward(self, x):
        return self.layers(x)

    def act(self, state, epsilon):
      if random.random() > epsilon:
          if not isinstance(state, torch.Tensor):
              state = torch.FloatTensor(np.array(state)).to(device)
          if state.dim() == 1:
              state = state.unsqueeze(0)

          with torch.no_grad():
              q_value = self.forward(state)
          action = q_value.max(1)[1].item()
      else:
          action = random.randrange(env.action_space.n)
      return action
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = DQN(env.observation_space.shape[0], env.action_space.n).to(device)
target_model = DQN(env.observation_space.shape[0], env.action_space.n).to(device)

# Синхронизация весов
target_model.load_state_dict(model.state_dict())

optimizer = optim.Adam(model.parameters())
```

<h2>Computing Temporal Difference Loss</h2>

```python
def compute_td_loss(batch_size):
    state, action, reward, next_state, done = replay_buffer.sample(batch_size)

    state      = torch.FloatTensor(np.float32(state)).to(device)
    next_state = torch.FloatTensor(np.float32(next_state)).to(device)
    action     = torch.LongTensor(action).to(device)
    reward     = torch.FloatTensor(reward).to(device)
    done       = torch.FloatTensor(done).to(device)

    q_values      = model(state)
    next_q_values = target_model(next_state)

    q_value          = q_values.gather(1, action.unsqueeze(1)).squeeze(1)
    next_q_value     = next_q_values.max(1)[0]
    expected_q_value = reward + gamma * next_q_value * (1 - done)

    loss = (q_value - expected_q_value.detach()).pow(2).mean()

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    return loss
```

```python
def plot(frame_idx, rewards, losses):
    clear_output(True)
    plt.figure(figsize=(20, 5))
    plt.subplot(131)
    mean_reward = np.mean(rewards[-10:]) if rewards else 0
    plt.title('frame %s. reward: %s' % (frame_idx, mean_reward))
    plt.plot(rewards)
    plt.ylim(0, 300)

    plt.subplot(132)
    plt.title('loss')
    plt.plot(losses)

    plt.show()
```

<h2>Training</h2>

```python
num_frames = 30000
batch_size = 32
gamma      = 0.99

losses = []
all_rewards = []
episode_reward = 0

res = env.reset()
state = res[0] if isinstance(res, tuple) else res

for frame_idx in range(1, num_frames + 1):
    epsilon = epsilon_by_frame(frame_idx)

    # Метод act сам разберется с тензором
    action = model.act(state, epsilon)

    step_res = env.step(action)
    if len(step_res) == 5:
        next_state, reward, done, truncated, _ = step_res
        done = done or truncated
    else:
        next_state, reward, done, _ = step_res

    replay_buffer.push(state, action, reward, next_state, done)

    state = next_state
    episode_reward += reward

    if done:
        res = env.reset()
        state = res[0] if isinstance(res, tuple) else res
        all_rewards.append(episode_reward)
        episode_reward = 0

    if len(replay_buffer) > batch_size:
        loss = compute_td_loss(batch_size)
        losses.append(loss.item())

    if frame_idx % 200 == 0:
        plot(frame_idx, all_rewards, losses)

    if frame_idx % 1000 == 0:
        target_model.load_state_dict(model.state_dict())
```

```python
def compute_td_loss(batch_size, beta):
    state, action, reward, next_state, done, indices, weights = replay_buffer.sample(batch_size, beta)

    state      = Variable(torch.FloatTensor(np.float32(state)))
    next_state = Variable(torch.FloatTensor(np.float32(next_state)))
    action     = Variable(torch.LongTensor(action))
    reward     = Variable(torch.FloatTensor(reward))
    done       = Variable(torch.FloatTensor(done))
    weights    = Variable(torch.FloatTensor(weights))

    q_values      = current_model(state)
    next_q_values = target_model(next_state)

    q_value          = q_values.gather(1, action.unsqueeze(1)).squeeze(1)
    next_q_value     = next_q_values.max(1)[0]
    expected_q_value = reward + gamma * next_q_value * (1 - done)

    diff = expected_q_value.detach() - q_value
    prios = diff.abs() + 1e-5
    loss  = (diff).pow(2) * weights
    loss  = loss.mean()

    optimizer.zero_grad()
    loss.backward()
    replay_buffer.update_priorities(indices, prios.data.cpu().numpy())
    optimizer.step()

    return loss
```

```python
def update_target(current_model, target_model):
    target_model.load_state_dict(current_model.state_dict())
```

```python
beta_start = 0.4
beta_frames = 1000
beta_by_frame = lambda frame_idx: min(1.0, beta_start + frame_idx * (1.0 - beta_start) / beta_frames)

replay_buffer = NaivePrioritizedBuffer(100000)
```

```python
class NaivePrioritizedBuffer(object):
    def __init__(self, capacity, prob_alpha=0.6):
        self.prob_alpha = prob_alpha
        self.capacity   = capacity
        self.buffer     = []
        self.pos        = 0
        self.priorities = np.zeros((capacity,), dtype=np.float32)

    def push(self, state, action, reward, next_state, done):
        state      = np.expand_dims(state, 0)
        next_state = np.expand_dims(next_state, 0)

        max_prio = self.priorities[:len(self.buffer)].max() if self.buffer else 1.0

        if len(self.buffer) < self.capacity:
            self.buffer.append((state, action, reward, next_state, done))
        else:
            self.buffer[self.pos] = (state, action, reward, next_state, done)

        self.priorities[self.pos] = max_prio
        self.pos = (self.pos + 1) % self.capacity

    def sample(self, batch_size, beta=0.4):
        N = len(self.buffer)
        prios = self.priorities[:N]

        probs  = prios ** self.prob_alpha
        probs /= probs.sum()

        indices = np.random.choice(N, batch_size, p=probs)
        samples = [self.buffer[idx] for idx in indices]

        weights  = (N * probs[indices]) ** (-beta)
        weights /= weights.max()
        weights  = np.array(weights, dtype=np.float32)

        batch       = list(zip(*samples))
        states      = np.concatenate(batch[0])
        actions     = batch[1]
        rewards     = batch[2]
        next_states = np.concatenate(batch[3])
        dones       = batch[4]

        return states, actions, rewards, next_states, dones, indices, weights

    def update_priorities(self, batch_indices, batch_priorities):
        for idx, prio in zip(batch_indices, batch_priorities):
            self.priorities[idx] = prio

    def __len__(self):
        return len(self.buffer)
```

```python
current_model = DQN(env.observation_space.shape[0], env.action_space.n)
target_model  = DQN(env.observation_space.shape[0], env.action_space.n)

optimizer = optim.Adam(current_model.parameters())

replay_buffer = NaivePrioritizedBuffer(1000)
update_target(current_model, target_model)
```
