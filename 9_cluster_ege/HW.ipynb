```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from google.colab import drive

drive.mount('/content/drive')
df = pd.read_excel('/content/drive/MyDrive/MachineLearning/8_cluster_ege/0.xlsx')
X = df.values

plt.scatter(X[:, 0], X[:, 1], s=10)
plt.title("Исходное распределение")
plt.show()

n_clusters = 2
model = KMeans(n_clusters=n_clusters, random_state=42).fit(X)
labels = model.labels_

plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=10)
plt.title("Кластеры")
plt.show()

def find_medoid(points):
    from scipy.spatial.distance import cdist
    dists = cdist(points, points, metric='euclidean')
    return points[np.argmin(dists.sum(axis=1))]

centers = []
for i in range(n_clusters):
    cluster_points = X[labels == i]
    medoid = find_medoid(cluster_points)
    centers.append(medoid)
    print(f"Медиана кластера {i}: {medoid}")

avg_center = np.mean(centers, axis=0)
print(f"Результат (Усреднение * 1000): {avg_center * 1000}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from google.colab import drive

drive.mount('/content/drive')
df = pd.read_excel('/content/drive/MyDrive/MachineLearning/8_cluster_ege/5.xlsx')
X = df.values

labels = []
for p in X:
    x, y = p[0], p[1]
    if (y >= 3.5 and x <= 0) or (y > x + 4 and x >= 0):
        labels.append(0)
    elif x <= 0:
        labels.append(1)
    else:
        labels.append(2)

labels = np.array(labels)

plt.figure(figsize=(8, 6))
colors = ['blue', 'red', 'green']
for i in range(3):
    plt.scatter(X[labels==i, 0], X[labels==i, 1], c=colors[i], s=10, label=f'Кластер {i}')

plt.title("Кластеры")
plt.legend()
plt.show()

def find_medoid(points):
    dists = cdist(points, points, metric='euclidean')
    return points[np.argmin(dists.sum(axis=1))]

centers = []
for i in range(3):
    cluster_points = X[labels == i]
    if len(cluster_points) > 0:
        medoid = find_medoid(cluster_points)
        centers.append(medoid)
        print(f"Центр кластера {i}: {medoid}")

avg_center = np.mean(centers, axis=0)
print(f"\nИТОГОВЫЙ ОТВЕТ (Усреднение * 1000): {int(avg_center[0]*1000)} {int(avg_center[1]*1000)}")
```
